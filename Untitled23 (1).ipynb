```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
df = pd.read_csv(r"/content/CC GENERAL.csv")
df.head()
```

```python

print(df.shape)
print(df.isnull().sum())
```

```python
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(df['MINIMUM_PAYMENTS'].mean())
df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(df['CREDIT_LIMIT'].median())
```

```python
df.isnull().sum()
```

```python
X = df.drop("CUST_ID", axis=1)
```

```python
plt.scatter(X.iloc[:, 0],X.iloc[:, 1])
plt.show()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 11), inertia)
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
plt.scatter(X_scaled[:, 0],X_scaled[:, 1],c=kmeans_labels)
plt.show()
```

```python
dbscan = DBSCAN(eps=2, min_samples=10)
dbscan_labels = dbscan.fit_predict(X_scaled)
print(set(dbscan_labels))
```

```python

plt.scatter(X_scaled[:, 0],X_scaled[:, 1],c=dbscan_labels)
plt.show()
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
pca_df = pd.DataFrame(X_pca,columns=["PC1", "PC2"])
```

```python

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    c=kmeans_labels
)
plt.show()
```

```python

plt.scatter(pca_df["PC1"],pca_df["PC2"],c=dbscan_labels)
plt.show()
```

```python
df['KMeans_Cluster'] = kmeans_labels
df['DBSCAN_Cluster'] = dbscan_labels
```

```python
df[['KMeans_Cluster', 'DBSCAN_Cluster']].head()
```

```python
cluster_summary = df.groupby('KMeans_Cluster')[
    [
        'BALANCE',
        'PURCHASES',
        'CASH_ADVANCE',
        'CREDIT_LIMIT',
        'PAYMENTS',
        'PURCHASES_FREQUENCY',
        'PRC_FULL_PAYMENT'
    ]
].mean().round(2)

cluster_summary
```

```python
cluster_summary = (
    df.drop(columns=["CUST_ID", "DBSCAN_Cluster"])
      .groupby("KMeans_Cluster")
      .mean()
)
cluster_summary
```

Cluster 0: Customers in this cluster have relatively low balances and low purchasing activity, representing lower-usage credit card customers.
Cluster 1: These customers show higher purchase amounts and purchase frequency, suggesting they are active credit card users.
Cluster 2: Customers have relatively high balances and cash advances, indicating heavier reliance on credit and cash advances.
Cluster 3: These customers have higher purchases, payments, and credit limits, representing high-value or high-spending customers.
DBSCAN: DBSCAN forms clusters based on dense regions and labels unusual customers as -1 (noise). Unlike K-Means, it does not require us to specify the number of clusters beforehand.
